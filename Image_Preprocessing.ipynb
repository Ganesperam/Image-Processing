```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("/content/processing_image.png")
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
```

IMAGE RESIZE

```python
print("Original size",img.shape)
img_resized = cv2.resize(img,(150,150))
print("updated size",img_resized.shape)
```

IMAGE RESHAPE

```python
img_reshaped = img_resized.reshape(-1, 3)
print("Reshaped image dimensions:", img_reshaped.shape)
```

CONVERT TO GRAY SCALE

```python
gray = cv2.cvtColor(img_resized,cv2.COLOR_BGR2GRAY)
plt.imshow(gray,cmap="gray")
plt.title("Grayscale Image")
plt.show()
```

NORMALIZATION

```python
img_normalized =img_resized / 355.0
```

```python
img_normalized
```

BLUR IMAGE

```python
blur = cv2.GaussianBlur(img_resized, (5,5),0)
plt.imshow(blur)
```

CANNY IMAGE

```python
edges = cv2.Canny(gray,100,200)

plt.imshow(edges, cmap="gray")
plt.title("Edge Detection")
plt.show()
```
